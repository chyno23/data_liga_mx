```python
import pandas as pd
import numpy as np

def extr_url_seriea(urlpath):

    data = pd.read_html('https://fbref.com/es/comps/11/2019-2020/horario/Marcadores-y-partidos-de-2019-2020-Serie-A')
    data = data[0]
    for i in range(0,5):
        urlname = '202'+str(i)+'-'+'202'+str(i+1)+'/horario/Marcadores-y-partidos-de-202'+str(i)+'-202'+str(i+1)+'-Serie-A'
        url = urlpath + urlname 

        sesons_data = pd.read_html(url)
        sesons_data = sesons_data[0]
        data = pd.concat([data, sesons_data],axis=0)
    return data
```

```python
import re
def extraer_numero(cadena):
    numeros = re.findall(r'\b\d+\b', str(cadena))
    if numeros:
        return int(numeros[0])
    else:
        return None
```

```python
urlpath = 'https://fbref.com/es/comps/11/'
data = extr_url_seriea(urlpath)
```

```python
data = data.drop('Notas',axis=1)
```

```python
data = data.drop('Ronda',axis=1)
data
```

```python
data[['Goles Local', 'Goles Visitante']] = data['Marcador'].str.split('–', expand=True)
data = data.drop('Marcador', axis=1)
data['Goles Local'] = data['Goles Local'].apply(extraer_numero).fillna(0).astype(int)
data['Goles Visitante'] = data['Goles Visitante'].apply(extraer_numero).fillna(0).astype(int)
data = data.reset_index(drop=True)
```

```python
data = data.dropna(subset=['Árbitro'])
```
