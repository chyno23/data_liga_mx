```python
data = pd.read_html('https://fbref.com/es/comps/9/2022-2023/horario/Marcadores-y-partidos-de-2022-2023-Premier-League',encoding='UTF-8')
data = data[0]
data[['Goles Local', 'Goles Visitante']] = data['Marcador'].str.split('–', expand=True)
data = data.drop('Marcador', axis=1)
data['Goles Local'] = data['Goles Local'].apply(extraer_numero).fillna(0).astype(int)
data['Goles Visitante'] = data['Goles Visitante'].apply(extraer_numero).fillna(0).astype(int)
data = data.reset_index(drop=True)
equipos = pd.concat([data['Local'],data['Visitante']]).dropna()
equipos_unicos = equipos.unique()
equipos_unicos_ordenados = sorted(equipos_unicos)
equipos_unicos_ordenados
```

```python
data = data.drop('Notas',axis=1)
data = data.drop('xG',axis=1)
data = data.drop('xG.1',axis=1)
data[:20]
```

```python
# Goles a favor como local y visitante
goles_favor_local = data.groupby("Local")["Goles Local"].sum()
goles_favor_visitante = data.groupby("Visitante")["Goles Visitante"].sum()
goles_a_favor = goles_favor_local.add(goles_favor_visitante, fill_value=0)

# Goles en contra como local y visitante
goles_contra_local = data.groupby("Local")["Goles Visitante"].sum()
goles_contra_visitante = data.groupby("Visitante")["Goles Local"].sum()
goles_en_contra = goles_contra_local.add(goles_contra_visitante, fill_value=0)
```
