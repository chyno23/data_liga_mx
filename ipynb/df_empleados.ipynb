```python
import requests

# Clave de la API (debes obtenerla registrándote en la API)
API_KEY = 'd527519e11msh7ce963ccbe620b9p10158ejsn257bb56b98a2'  # Reemplaza con tu API Key

# URL de la API para obtener partidos en vivo
url_partidos = "https://v3.football.api-sports.io/fixtures?live=all"
headers = {
    'x-rapidapi-key': API_KEY,
    'x-rapidapi-host': 'api-football-v1.p.rapidapi.com'
}

# Hacer la solicitud a la API para obtener partidos en vivo
response_partidos = requests.get(url_partidos, headers=headers)
data_partidos = response_partidos.json()

# Verificar si la solicitud fue exitosa
if response_partidos.status_code == 200:
    print("Solicitud exitosa")
    if data_partidos['response']:
        # Iterar sobre cada partido en vivo
        lista_de_partidos = []
        for partido in data_partidos['response']:
            partido_id = partido['fixture']['id']
            equipo_local = partido['teams']['home']['name']
            equipo_visitante = partido['teams']['away']['name']
            marcador_local = partido['goals']['home']
            marcador_visitante = partido['goals']['away']

            lista_de_partidos.append({
                "Partido_ID": partido_id,
                "Equipo_Local": equipo_local,
                "Equipo_Visitante": equipo_visitante,
                "Goles_Local": marcador_local,
                "Goles_Visitante": marcador_visitante
            })

            # Obtener eventos del partido
            url_eventos = f'https://api-football-v1.p.rapidapi.com/v3/fixtures/events?search={partido_id}'
            response_eventos = requests.get(url_eventos, headers=headers)
            data_eventos = response_eventos.json()

            # Verificar si la solicitud de eventos fue exitosa
            if response_eventos.status_code == 200 and data_eventos['response']:
                # Filtrar tarjetas rojas
                tarjetas_rojas = []
                for evento in data_eventos['response']:
                    if evento.get('type') == 'Card' and evento.get('detail') == 'Red Card':
                        equipo = evento['team']['name']
                        jugador = evento['player']['name'] if evento['player'] else 'Desconocido'
                        tiempo = evento['time']['elapsed']
                        tarjetas_rojas.append((equipo, jugador, tiempo))

                # Mostrar tarjetas rojas
                if tarjetas_rojas:
                    print("Tarjetas rojas en este partido:")
                    for equipo, jugador, tiempo in tarjetas_rojas:
                        print(f"  - {equipo}: {jugador} (minuto {tiempo}')")
                else:
                    print("  No hay tarjetas rojas en este partido.")
            else:
                print("  No se pudieron obtener los eventos del partido.")
    else:
        print("No hay partidos en vivo en este momento.")
else:
    print(f"Error en la solicitud: {response_partidos.status_code}")
    data_partidos  # Imprimir la respuesta para ver el mensaje de error

df_futbol = pd.DataFrame(lista_de_partidos)
df_futbol
```

```python
import requests
from bs4 import BeautifulSoup

# URL de ESPN con la lista de partidos próximos
url = "https://www.espn.com/soccer/scoreboard"

# Hacer la solicitud
response = requests.get(url)
soup = BeautifulSoup(response.text, "html.parser")

# Buscar los contenedores de partidos
partidos = soup.find_all("article", class_="Scoreboard__item")

# Extraer información de cada partido
for partido in partidos:
    equipos = partido.find_all("span", class_="ScoreCell__TeamName")
    fecha_hora = partido.find("span", class_="ScoreCell__Time")

    if equipos and fecha_hora:
        equipo_local = equipos[0].text.strip()
        equipo_visitante = equipos[1].text.strip()
        horario = fecha_hora.text.strip()
        
        print(f"{equipo_local} vs {equipo_visitante} - {horario}")
```
