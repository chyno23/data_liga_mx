```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt 
```

```python
import re
def extr_url_primer(urlpath):

    data = pd.read_html('https://fbref.com/es/comps/20/2019-2020/horario/Marcadores-y-partidos-de-2019-2020-Bundesliga')
    data = data[0]
    for i in range(0,5):
        urlname = '202'+str(i)+'-'+'202'+str(i+1)+'/horario/Marcadores-y-partidos-de-202'+str(i)+'-202'+str(i+1)+'-Bundesliga'
        url = urlpath + urlname 

        sesons_data = pd.read_html(url)
        sesons_data = sesons_data[0]
        data = pd.concat([data, sesons_data],axis=0)
    return data

def extraer_numero(cadena):
    numeros = re.findall(r'\b\d+\b', str(cadena))
    if numeros:
        return int(numeros[0])
    else:
        return None
```

```python
urlpath = 'https://fbref.com/es/comps/9/'
data = extr_url_primer(urlpath)
```
