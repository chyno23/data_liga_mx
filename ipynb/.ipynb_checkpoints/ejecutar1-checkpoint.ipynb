```python
import pandas as pd
import numpy as np

def extr_url_primer(urlpath):

    data = pd.read_html('https://fbref.com/es/comps/9/2019-2020/horario/Marcadores-y-partidos-de-2019-2020-Premier-League')
    data = data[0]
    for i in range(0,5):
        urlname = '202'+str(i)+'-'+'202'+str(i+1)+'/horario/Marcadores-y-partidos-de-202'+str(i)+'-202'+str(i+1)+'-Premier-League'
        url = urlpath + urlname 

        sesons_data = pd.read_html(url)
        sesons_data = sesons_data[0]
        data = pd.concat([data, sesons_data],axis=0)
    return data
```

```python
urlpath = 'https://fbref.com/es/comps/9/'
```

```python
data = extr_url_primer(urlpath)
```

```python
import re
def extraer_numero(cadena):
    numeros = re.findall(r'\b\d+\b', str(cadena))
    if numeros:
        return int(numeros[0])
    else:
        return None
```

```python
data[['Goles Local', 'Goles Visitante']] = data['Marcador'].str.split('–', expand=True)
data = data.drop('Marcador', axis=1)
data['Goles Local'] = data['Goles Local'].apply(extraer_numero).fillna(0).astype(int)
data['Goles Visitante'] = data['Goles Visitante'].apply(extraer_numero).fillna(0).astype(int)
data = data.reset_index(drop=True)
```

```python
# Sumar los goles de local y visita para cada equipo
goles_local = data.groupby('Local')['Goles Local'].sum().reset_index()
goles_visitante = data.groupby('Visitante')['Goles Visitante'].sum().reset_index()

# Renombrar las columnas para identificar los goles de local y visitante
goles_local.rename(columns={'Goles Local': 'Goles'}, inplace=True)
goles_visitante.rename(columns={'Goles Visitante': 'Goles'}, inplace=True)

# Unir los goles de local y visitante para obtener el total
goles_totales = pd.concat([goles_local, goles_visitante]).groupby('Equipo')['Goles'].sum().reset_index()
```

```python
equipos_unicos = pd.unique(data[['Local', 'Visitante']].values.ravel())
# Crear un DataFrame con todos los equipos
equipos_data = pd.DataFrame(equipos_unicos, columns=['Equipo'])

# Calcular los goles de local
goles_local = data.groupby('Local')['Goles Local'].sum().reset_index().rename(columns={'Local': 'Equipo', 'Goles Local': 'Goles'})

# Calcular los goles de visitante
goles_visitante = data.groupby('Visitante')['Goles Visitante'].sum().reset_index().rename(columns={'Visitante': 'Equipo', 'Visitante': 'Goles'})

# Unir los goles de local y visitante
goles_totales = pd.concat([goles_local, goles_visitante]).groupby('Equipo')['Goles'].sum().reset_index()

# Asegurarse de que todos los equipos estén presentes, incluso aquellos sin goles
equipos_data = pd.merge(equipos_data, goles_totales, on='Equipo', how='left').fillna(0)
```

```python
equipos_data.sort_values(by='Goles',ascending=False).reset_index(drop=True)
```

```python
partidos_local = data.groupby('Local').size().reset_index(name='Partidos Jugados Local')
```

```python
partidos_totales['Partidos Totales'] = partidos_totales['Partidos_Jugados_Local'] + partidos_totales['Partidos_Jugados_Visitante']
```

```python
# Contar los partidos jugados como local para cada equipo
partidos_local = data.groupby('Local').size().reset_index(name='Partidos Local')

# Contar los partidos jugados como visitante para cada equipo
partidos_visitante = data.groupby('Visitante').size().reset_index(name='Partidos Visitante')

# Unir las estadísticas de partidos jugados como local y visitante
partidos_totales = pd.merge(partidos_local, partidos_visitante, left_on='Local', right_on='Visitante', how='outer')

# Renombrar las columnas para unificar
partidos_totales = partidos_totales.rename(columns={'Local': 'Equipo', 'Partidos Local': 'Partidos Jugados Local', 'Partidos Visitante': 'Partidos Jugados Visitante'})

# Calcular el total de partidos jugados por cada equipo (local + visitante)
partidos_totales['Partidos Totales'] = partidos_totales['Partidos Jugados Local'] + partidos_totales['Partidos Jugados Visitante']
```

```python
partidos_totales=partidos_totales.drop('Partidos Jugados Local',axis=1)
partidos_totales=partidos_totales.drop('Visitante',axis=1)
partidos_totales=partidos_totales.drop('Partidos Jugados Visitante',axis=1)
```

```python
partidos_totales.sort_values(by='Partidos Totales',ascending=False).reset_index(drop=True)
```
