```python
import pandas as pd
import matplotlib.pyplot as pyplot
import numpy as np
```

```python
jugador = pd.read_html('https://fbref.com/es/comps/31/shooting/Estadisticas-de-Liga-MX', encoding='UTF-8')
```

```python
data = jugador[0].iloc[:,:]
```

```python
data.to_csv('jugadores_mx.csv')
```

```python
data = pd.read_csv('jugadores_mx.csv')
```

```python
data.sort_values(by='Gls.', ascending=False)
```

```python
from sklearn.preprocessing import MinMaxScaler

# Step 1: Drop the unnecessary column ('Unnamed: 0') and any rows with missing values
jugadores_df_cleaned = data.drop(columns=['Unnamed: 0']).dropna()

# Step 2: Standardize column names to remove special characters and spaces
jugadores_df_cleaned.columns = jugadores_df_cleaned.columns.str.replace(r'[^A-Za-z0-9_]', '_', regex=True)

# Step 3: Scale numerical columns using MinMaxScaler
# Select only numerical columns for scaling
numeric_columns = jugadores_df_cleaned.select_dtypes(include=['float64']).columns
scaler = MinMaxScaler()
jugadores_df_cleaned[numeric_columns] = scaler.fit_transform(jugadores_df_cleaned[numeric_columns])

# Display the cleaned and normalized data
jugadores_df_cleaned
```

```python
jugadores_df_cleaned.to_csv('jugadores_mx_normalize.csv')
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from fpdf import FPDF

# Leer los datos normalizados
data = pd.read_csv('jugadores_mx_normalize.csv')

# Crear un archivo PDF
pdf = FPDF()
pdf.add_page()

# 1. Agregar título
pdf.set_font("Arial", 'B', 16)
pdf.cell(200, 10, "Reporte de Datos Normalizados de Jugadores", 0, 1, 'C')

# 2. Estadísticas descriptivas
pdf.set_font("Arial", 'B', 12)
pdf.cell(200, 10, "Estadísticas Descriptivas", 0, 1, 'L')
pdf.set_font("Arial", '', 10)

# Convertir a tabla y agregar al PDF
desc_stats = data.describe().round(2)
pdf.cell(200, 10, str(desc_stats), 0, 1, 'L')

# 3. Gráficas de Distribución
# Crear las figuras y guardarlas como imágenes
fig, ax = plt.subplots(figsize=(10, 6))
sns.histplot(data['Gls_'], kde=True, ax=ax)  # Cambiar 'ColumnaEjemplo' a la columna deseada
plt.savefig('distribucion.png')
pdf.image('distribucion.png', x=10, y=None, w=100)

# 4. Matriz de Correlación
plt.figure(figsize=(10, 8))
corr = data.corr()
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.savefig('correlacion.png')
pdf.add_page()
pdf.image('correlacion.png', x=10, y=None, w=100)

# 5. Guardar el PDF
pdf.output("reporte_jugadores_mx.pdf")
```
