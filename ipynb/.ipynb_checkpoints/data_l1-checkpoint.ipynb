```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

# Datos proporcionados
data = [
    [9, 836], [25, 838], [41, 835], [58, 833], [74, 833], [91, 831],
    [108, 832], [124, 819], [142, 839], [158, 858], [176, 838], [192, 844],
    [209, 838], [226, 836], [243, 838], [260, 844], [277, 827], [293, 837],
    [311, 830], [327, 831], [345, 858], [361, 854], [378, 820], [396, 835],
    [412, 837], [430, 838], [446, 838], [463, 840], [480, 843], [497, 840],
    [514, 839], [531, 831], [547, 847], [565, 867], [581, 851], [599, 847],
    [615, 847], [632, 844], [649, 842], [666, 843], [683, 844], [700, 843],
    [716, 844], [734, 843], [750, 829], [768, 858], [784, 834], [801, 855],
    [818, 847], [835, 845], [852, 837], [869, 843], [886, 842], [903, 843],
    [920, 843], [936, 843], [954, 842], [970, 843], [988, 849], [1004, 860],
    [1021, 827], [1039, 847], [1056, 847], [1074, 843], [1091, 846], [1107, 847],
    [1125, 844], [1142, 833], [1160, 839], [1177, 833], [1195, 827], [1211, 823],
    [1228, 854], [1246, 846], [1263, 821], [1281, 828], [1298, 833], [1315, 842],
    [1332, 842], [1349, 840], [1367, 843], [1384, 842], [1401, 839], [1419, 853],
    [1436, 838], [1453, 865], [1470, 849], [1487, 823], [1505, 846], [1522, 847],
    [1540, 845], [1556, 848], [1573, 845], [1591, 848], [1608, 843], [1626, 845],
    [1643, 828], [1660, 853], [1677, 839], [1694, 831]
]

# Convertimos los datos en arrays de numpy
data = np.array(data)
x = data[:, 0]
y = data[:, 1]

# Definimos la función senoidal
def sin_func(x, A, B, C, D):
    return A * np.sin(B * x + C) + D

# Estimamos los parámetros iniciales para el ajuste
A_initial = (max(y) - min(y)) / 2
B_initial = 2 * np.pi / (max(x) - min(x))
C_initial = 0
D_initial = np.mean(y)

# Ajustamos la función senoidal a los datos
initial_guess = [A_initial, B_initial, C_initial, D_initial]
params, params_covariance = curve_fit(sin_func, x, y, p0=initial_guess)

# Generamos puntos para la curva ajustada
x_fit = np.linspace(min(x), max(x), 1000)
y_fit = sin_func(x_fit, *params)

# Visualizamos los datos y la función ajustada
plt.scatter(x, y, label='Datos de muestra')
plt.plot(x_fit, y_fit, color='red', label='Ajuste senoidal')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.show()

# Imprimir los parámetros ajustados
print("Parámetros ajustados:")
print(f"A (Amplitud) = {params[0]}")
print(f"B (Frecuencia) = {params[1]}")
print(f"C (Desfase) = {params[2]}")
print(f"D (Desplazamiento vertical) = {params[3]}")
```
